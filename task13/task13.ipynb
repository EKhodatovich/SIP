```python
import cv2
from deepface import DeepFace

# Загрузка изображения
image_path = 'faces1.png'  # Путь к изображению

# Загружаем изображение с помощью OpenCV
image = cv2.imread(image_path)

# Преобразуем в серый цвет для лучшего обнаружения лиц
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Загружаем предварительно обученный классификатор для распознавания лиц
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Обнаруживаем лица на изображении
faces = face_cascade.detectMultiScale(gray, 1.1, 4)

# Для каждого обнаруженного лица анализируем возраст и эмоцию
for (x, y, w, h) in faces:
    # Вырезаем область лица для анализа
    face = image[y:y + h, x:x + w]

    # Анализируем возраст и эмоции с помощью DeepFace
    try:
        analysis = DeepFace.analyze(face, actions=['age', 'emotion'])
        
        # Если возвращается список, берем первый элемент
        if isinstance(analysis, list):
            analysis = analysis[0]
        
        # Получаем возраст и доминирующую эмоцию
        age = analysis['age']
        dominant_emotion = analysis['dominant_emotion']
        
        # Подготавливаем текст для подписи
        label = f'Age: {age}, Emotion: {dominant_emotion}'
        
        # Рисуем рамку вокруг лица
        cv2.rectangle(image, (x, y), (x + w, y + h), (0, 255, 0), 2)
        
        # Выводим текст с возрастом и эмоцией над рамкой
        cv2.putText(image, label, (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)
    
    except Exception as e:
        print(f"Ошибка при анализе лица: {e}")

# Отображаем результат
cv2.imshow('Image with Face Detection and Analysis', image)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Сохраняем изображение с результатами
cv2.imwrite('result.jpg', image)
```
